```python
# PAI PROJECT
# 22k-4018
# 22k-4082
# 22k-4059

import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
dataframe=pd.read_csv(r'D:\Sem 3\PAI\Project\Adidas US Sales Datasets.csv', skiprows= 3)
dataframe=pd.DataFrame(dataframe)

```

```python
no_of_products=list(dataframe['Product'].unique())
print("In this file record of " + str(len(no_of_products)) + " products are present: ")
print(no_of_products)
```

```python
no_of_retailer=list(dataframe['Retailer'].unique())
print("In this file record of " + str(len(no_of_retailer)) + " retailer are present: ")
print(no_of_retailer)
```

```python
no_of_States = list(dataframe['State'].unique())
print("In this file record of " + str(len(no_of_States)) + " States data is present: ")
print(no_of_States)
```

```python
no_of_Region = list(dataframe['Region'].unique())
print("In this file record of " + str(len(no_of_Region)) + " region are present: ")
print(no_of_Region)
```

```python
no_of_Cities = list(dataframe['City'].unique())
print("In this file, record of " + str(len(no_of_Cities)) + " cities is present which are: ")
print(no_of_Cities)
```

```python
no_of_sales_method=list(dataframe['Sales Method'].unique())
print("In this file, record of " + str(len(no_of_sales_method)) + " sales methods are: ")
print(no_of_sales_method)
```

```python
# Total Units sold in different States

total_sales_forEvery_State = {} # Lengthy name but ez to understand:)

# if we remove this line, Units Sold is treated as a string which gives us wrong answers.
dataframe['Units Sold'] = pd.to_numeric(dataframe['Units Sold'], errors='coerce')

# Take out what we need
for state in no_of_States:
    selected = dataframe['State'] == state
    total_sales_forEvery_State[state] = dataframe.loc[selected, 'Units Sold'].sum()
    
for state, total_units_sold in total_sales_forEvery_State.items():
 print(f"Total units sold in {state}: {total_units_sold}")
```

```python
# Total Units sold in different cities

total_sales_forEvery_City = {}

# if we remove this line, Units Sold is treated as a string which gives us wrong answers.
dataframe['Units Sold'] = pd.to_numeric(dataframe['Units Sold'], errors='coerce')

for city in no_of_Cities:
    selected = dataframe['City'] == city
    total_sales_forEvery_City[city] = dataframe.loc[selected, 'Units Sold'].sum()
    
for city, total_units_sold in total_sales_forEvery_City.items():
    print(f"Total units sold in {city}: {total_units_sold}")
```

```python
# Average Sale of Every City

total_avg_sales_forEvery_City = {}

# if we remove this line, Units Sold is treated as a string which gives us wrong answers.
dataframe['Units Sold'] = pd.to_numeric(dataframe['Units Sold'], errors='coerce')

for city in no_of_Cities:
    selected = dataframe['City'] == city
    total_avg_sales_forEvery_City[city] = dataframe.loc[selected, 'Units Sold'].mean()
    
for city, avg_units_sold in total_avg_sales_forEvery_City.items():
    print(f"Total units sold in {city}: {avg_units_sold}")
```

```python
# Average Sale of every State

total_avg_sales_forEvery_State = {}

# if we remove this line, Units Sold is treated as a string which gives us wrong answers.
dataframe['Units Sold'] = pd.to_numeric(dataframe['Units Sold'], errors='coerce')

for state in no_of_States:
    selected = dataframe['State'] == state
    total_avg_sales_forEvery_City[state] = dataframe.loc[selected, 'Units Sold'].mean()
    
for state, avg_units_sold in total_avg_sales_forEvery_City.items():
    print(f"Total units sold in {state}: {avg_units_sold}")
```

```python
# Plotting Total Unit Sales on a Bar Graph

KEYS = total_sales_forEvery_City.keys()
VALUES = total_sales_forEvery_City.values()

plt.figure(figsize=(15, 6))
plt.bar(KEYS, VALUES, color= 'lightblue')
plt.title("Unit Sales of Cities in US")
plt.xlabel('Cities')
plt.ylabel('Unit Sales')
plt.xticks(rotation= 90, ha= 'right', fontsize= 6)


plt.show()
```

```python
# Now diaplaying total sales of each region on a Pie Chart so that it can be read easily
from matplotlib import pyplot as plt

total_sales_region = {}

dataframe['Total Sales'] = dataframe['Total Sales'].replace('[\\$,]', '', regex= True).astype(float)

for region in no_of_Region:
    selected = dataframe['Region'] == region
    total_sales_region[region] = dataframe.loc[selected, 'Total Sales'].sum()
    
plt.figure(figsize=(5, 5), facecolor='None') 
plt.pie(total_sales_region.values(), labels=total_sales_region.keys(), autopct='%1.1f%%', startangle=140, textprops= {'color': 'white'})

plt.axis('Equal')
plt.title("Total Sales of every Region", color= 'White')
plt.show()
```

```python
#Units Sold of every Region
from matplotlib import pyplot as plt

unit_sold_region = {}

dataframe['Units Sold'] = dataframe['Units Sold'].replace('[\\$,]', '', regex= True).astype(float)

for region in no_of_Region:
    selected = dataframe['Region'] == region
    unit_sold_region[region] = dataframe.loc[selected, 'Units Sold'].sum()
    
plt.figure(figsize=(5, 5), facecolor='None') 
plt.pie(unit_sold_region.values(), labels=unit_sold_region.keys(), autopct='%1.1f%%', startangle=140, textprops= {'color': 'white'})

plt.axis('Equal')
plt.title("Units Sold of every Region", color= 'White')
plt.show()
```

```python
#Decesion tree of Total Sales and Operating Profit
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier 
from sklearn.model_selection import train_test_split 
from sklearn.metrics import accuracy_score  
from sklearn.metrics import confusion_matrix


dataframe['Total Sales'] = dataframe['Total Sales'].replace('[\\$,]', '', regex=True).astype(float)
dataframe['Operating Profit'] = dataframe['Operating Profit'].replace('[\\$,]', '', regex=True).astype(float) 

X = dataframe[['Total Sales']]  
Y = dataframe['Operating Profit']  


X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

clf = DecisionTreeClassifier()

clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")

```

```python
#Naive Bayes of Total Sales ans Operating Profit
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier 
from sklearn.naive_bayes import GaussianNB  
from sklearn.model_selection import train_test_split 
from sklearn.metrics import accuracy_score  


dataframe['Total Sales'] = dataframe['Total Sales'].replace('[\\$,]', '', regex=True).astype(float)
dataframe['Operating Profit'] = dataframe['Operating Profit'].replace('[\\$,]', '', regex=True).astype(float) 

X = dataframe[['Total Sales']]  
Y = dataframe['Operating Profit']  

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

clf_nb = GaussianNB()  
clf_nb.fit(X_train, y_train)
y_pred_nb = clf_nb.predict(X_test)
accuracy_nb = accuracy_score(y_test, y_pred_nb)
print(f"Naive Bayes Accuracy: {accuracy_nb:.2f}")
```

```python
# What is the best Sales Technique

sales_by_method = {}

for sales in no_of_sales_method:
    selected = dataframe['Sales Method'] == sales
    sales_by_method[sales] = dataframe.loc[selected, 'Units Sold'].sum()
    

# Geting the key of the method that has max sales.
max_sum = max(sales_by_method, key=sales_by_method.get)

# Exploding that area for better visuals.
Slice = [0.2 if method == max_sum else 0 for method in sales_by_method.keys()]

# Plotting it
plt.figure(figsize= (5, 5), facecolor='None')
plt.pie(sales_by_method.values(), labels=sales_by_method.keys(), autopct= '%1.1f%%', textprops= {'color': 'white'}, explode= Slice)
plt.axis('Equal')
plt.title('Best Sales Method', color= 'green')
plt.show()
print("Pie chart shows " + str(max_sum) + " selling has produced maximum sales.")
```

```python
# Which Retailer has most sales

most_sales_by_retailer = {}

for retailer in no_of_retailer:
    selected = dataframe['Retailer'] == retailer
    most_sales_by_retailer[retailer] = dataframe.loc[selected, 'Units Sold'].sum()
    
    
KEYS = most_sales_by_retailer.keys()
VALUES = most_sales_by_retailer.values()

# Geting the key of the retailer that has max sales.
max_sum = max(most_sales_by_retailer, key=most_sales_by_retailer.get)

plt.figure(figsize= (6, 3), facecolor= 'None')
plt.bar(KEYS, VALUES, color= 'lightblue')
plt.xticks(rotation= 45, ha= 'right', fontsize= 8)
plt.title('Most Sales by retailer')
plt.xlabel("Retailers")
plt.ylabel("Sales")
plt.show()
print("Graph shows " + str(max_sum) + " has done most sales.")
```
